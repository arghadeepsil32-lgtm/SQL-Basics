**Q10. What Are Constraints in SQL and Why Are They Used?**

Answer ▶

constraints are rules applied to columns or tables to limit the type of data that can be entered. They act as a "security guard" for your database, ensuring that the information remains accurate, reliable, and consistent.

**Why Are Constraints Used?**
Data Integrity: They prevent accidental entry of invalid data (e.g., leaving a name blank when it's required).

- Uniqueness: They ensure that specific values, like an employee_id, are never duplicated.

- Relationship Management: They maintain the links between tables, ensuring a sale can't be assigned to an employee who doesn't exist.
